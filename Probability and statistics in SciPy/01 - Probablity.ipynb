```python
from scipy.stats import randint, beta
import numpy as np
```

```python
# Example: Rolling a Dice

dice = randint(1,7)

# Calculate probability of getting a number greater than 4
probablity = dice.sf(4)

print(f"Probability of getting a number greater than 4: {probablity}")
print(f"Probability of getting a number greater than 4: {probablity * 100}%")
```

```python
# Probability of the Sun Rising Tomorrow

n = 10000 # Number of previous successful sunrises

# Beta distribution parameters
a = n + 1
b = 1

probability = beta.mean(a, b)

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

```python
# die - Getting an Even Number
n = 6
even_nums = [num for num in range(2,n+1,2)]

p = dice.pmf(2) + dice.pmf(4) + dice.pmf(6)

print("Probability of getting an even number:", p)
print("Percentage:", p * 100, "%")
```

```python
# getting and even number from 1 to n
n = 10
nums = list(range(1, n+1))
even_nums = [num for num in nums if num % 2 == 0]
even_nums

p = 0
for even in even_nums:
    p += dice.pmf(even)

print("Range: 1 to", n)
print("Even Numbers:", even_nums)
print("Probability:", p)
print("Percentage:", p * 100, "%")
```

# Condiional Probablity

```python
# probability of selecting exactly 2 red balls when selecting 3 balls from a bag containing 5 red and 5 blue balls.

from scipy.stats import hypergeom

# Population information
total_balls = 10
red_balls = 5
selected_balls = 3

# Probability of getting exactly 2 red balls
probability = hypergeom.pmf(
    2,
    total_balls,
    red_balls,
    selected_balls
)

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

# Theorem of Total Probability

```python
from scipy.stats import bernoulli

# Production probability
machine_A = 0.60
machine_B = 0.40

# Defective probabilities
defective_A = 0.05
defective_B = 0.10

# Total probability theorem
probability = (
    machine_A * defective_A +
    machine_B * defective_B
)

print("Probability of defective product:", probability)
print("Percentage:", probability * 100, "%")
```

# Bayes' Theorem

```python
from scipy.stats import bernoulli

# Prior probability of disease
p_disease = 0.01

# Probability of positive test if diseased
p_positive_disease = 0.99

# Probability of positive test if healthy
p_positive_healthy = 0.05

# Calculate likelihoods using Bernoulli distribution
likelihood_disease = bernoulli.pmf(1, p_positive_disease)
likelihood_healthy = bernoulli.pmf(1, p_positive_healthy)

# Calculate overall probability of positive test
p_positive = (
    likelihood_disease * p_disease +
    likelihood_healthy * (1 - p_disease)
)

# Apply Bayes' Theorem
posterior = (
    likelihood_disease * p_disease
) / p_positive

print("Prior Probability:", p_disease)
print("Probability of Positive Test:", p_positive)
print("Posterior Probability:", posterior)
print("Final Percentage:", posterior * 100, "%")
```

# Random Variable

```python
import pandas as pd

# discrete - Random Variable Using SciPy - – Dice Example
dice = randint(1,7)

values = list(range(1,7))

value = []
probablity = []

# Random Varibale
for v in values:
    value.append(v)
    probablity.append(
        dice.pmf(v)
    )

rv = {
    'value' : value,
    'probablity' : probablity
}

random_varibale = pd.DataFrame(rv)

random_varibale

total_p = random_varibale['probablity'].sum()

random_varibale
```

```python
# continuous - Random Variable Using SciPy - – Dice Example

from scipy.stats import norm

# Create a continuous random variable
X = norm(loc=170, scale=10)

# Calculate probability density at height 175
result = X.pdf(175)

print("Probability Density:", result)
```

```python
# For a continuous random variable, we calculate probabilities over intervals using the CDF.
from scipy.stats import norm

X = norm(loc=170, scale=10)

probability = X.cdf(180) - X.cdf(160)

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```
