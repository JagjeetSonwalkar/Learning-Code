# Step 1: Import libraries

```python
import numpy as np
from scipy import stats
```

# Step 2: Define the dataset

```python
# Version A
visitors_A = 1000
conversions_A = 100

# Version B
visitors_B = 1000
conversions_B = 130

# Significance level
alpha = 0.05
```

# Step 3: Calculate conversion rates

```python
# Conversion Rate = {Conversions}/{Visitors} * 100

conversion_rate_A = (conversions_A / visitors_A) * 100

conversion_rate_B = (conversions_B / visitors_B) * 100

print("Version A Conversion Rate:", conversion_rate_A, "%")
print("Version B Conversion Rate:", conversion_rate_B, "%")
```

Observation:
- Version A conversion rate = 10%.
- Version B conversion rate = 13%.
Version B has a 3 percentage-point higher conversion rate.

However, we cannot conclude that Version B is statistically better just by comparing percentages.

We need to perform Hypothesis Testing.

# Step 4: Define hypotheses

# Step 5: Calculate the Z-statistic

```python
# Calculate conversion proportions
p_A = conversions_A / visitors_A
p_B = conversions_B / visitors_B

# Calculate pooled proportion
pooled_p = (
    conversions_A + conversions_B
) / (
    visitors_A + visitors_B
)

# Calculate standard error
SE = np.sqrt(
    pooled_p * (1 - pooled_p) *
    (1/visitors_A + 1/visitors_B)
)

# Calculate Z-statistic
z_stat = (p_B - p_A) / SE

print("Z-statistic:", z_stat)
```

# Step 6: Calculate P-value

```python
p_value = stats.norm.sf(z_stat)

print("P-value:", p_value)
```

# Step 7: Make the final decision

```python
if p_value < alpha:
    print("Reject Null Hypothesis")
    print("Version B performs significantly better.")
else:
    print("Fail to Reject Null Hypothesis")
    print("No sufficient evidence that Version B is better.")
```
