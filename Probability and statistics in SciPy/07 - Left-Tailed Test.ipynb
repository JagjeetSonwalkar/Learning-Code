# Left-Tailed Test using SciPy

```python
import numpy as np
from scipy import stats

# Sample data
production_time = [18, 17, 19, 16, 18, 17, 19, 18, 16, 17]

# Given values
population_mean = 20
population_std = 2
alpha = 0.05

# Sample statistics
sample_mean = np.mean(production_time)
sample_size = len(production_time)

# Calculate Z-statistic
z_stat = (sample_mean - population_mean) / (
    population_std / np.sqrt(sample_size)
)

# Left-tailed P-value
p_value = stats.norm.cdf(z_stat)

print("Sample Mean:", sample_mean)
print("Z-statistic:", z_stat)
print("P-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject Null Hypothesis")
    print("Production time has significantly decreased.")
else:
    print("Fail to Reject Null Hypothesis")
```
