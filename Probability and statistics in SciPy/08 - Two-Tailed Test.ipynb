# Two-Tailed Test using SciPy

```python
import numpy as np
from scipy import stats

# Sample data
weight = [510, 520, 505, 515, 512,
          508, 518, 511, 507, 514]

# Given values
population_mean = 500
population_std = 10
alpha = 0.05

# Sample statistics
sample_mean = np.mean(weight)
sample_size = len(weight)

# Calculate Z-statistic
z_stat = (sample_mean - population_mean) / (
    population_std / np.sqrt(sample_size)
)

# Two-tailed P-value
p_value = 2 * stats.norm.sf(abs(z_stat))

print("Sample Mean:", sample_mean)
print("Z-statistic:", z_stat)
print("P-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject Null Hypothesis")
    print("Average product weight is significantly different.")
else:
    print("Fail to Reject Null Hypothesis")
    print("Not enough evidence of a difference.")
```
