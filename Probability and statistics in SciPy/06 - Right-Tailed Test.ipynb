# Right-Tailed Test using SciPy

# Step 1: Import libraries

```python
import numpy as np
from scipy import stats
```

# Step 2: Define sample data and hypothesis values

```python
tasks = [52, 55, 54, 56, 53, 58, 57, 54, 55, 56]

population_mean = 50
alpha = 0.05
```

# Step 3: Calculate Z-statistic

```python
sample_mean = np.mean(tasks)
population_std = 3
sample_size = len(tasks)

z_stat = (sample_mean - population_mean) / (
    population_std / np.sqrt(sample_size)
)

print("Z-statistic:", z_stat)
```

# Step 4: Calculate P-value

```python
p_value = stats.norm.sf(z_stat)

print("P-value:", p_value)
```

# Step 5: Make a decision

```python
if p_value < alpha:
    print("Reject Null Hypothesis")
    print("Average productivity has significantly increased.")
else:
    print("Fail to Reject Null Hypothesis")
```
