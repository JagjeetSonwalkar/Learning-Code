# Problem statement
#### A company claims that the average salary of its employees is ₹30,000.

```python
# We collect salary data from 10 employees:
salary = [28000, 32000, 29000, 31000, 35000,
          30000, 27000, 33000, 29000, 34000]
```

#### Our goal is to check whether the average salary is significantly different from ₹30,000.

# Step 1: Import libraries

```python
import numpy as np
from scipy import stats
```

# Step 2: Create sample data

```python
salary = [28000, 32000, 29000, 31000, 35000,
          30000, 27000, 33000, 29000, 34000]

claimed_mean = 30000
```

# Step 3: Define hypotheses

# Step 4: Perform the T-test

```python
t_stat, p_value = stats.ttest_1samp(
    salary,
    popmean=claimed_mean
)

print("T-statistic:", t_stat)
print("P-value    :", p_value)
```

# Step 5: Make the decision

```python
alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis")
    print("Average salary is significantly different.")
else:
    print("Fail to Reject Null Hypothesis")
    print("Not enough evidence of a difference.")
```

# Step 6: Understand the result

```python
Our P-value is approximately 0.1483.

Significance level: alpha = 0.05
Comparison: 0.1483 > 0.05
```
