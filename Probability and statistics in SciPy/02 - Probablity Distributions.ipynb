# PMF using SciPy

```python
# Example: Rolling a dice

dice = randint(1, 7)      # Create a discrete uniform distribution
probability = dice.pmf(4) # Calculate PMF

print(probability)
```

```python
# Calculate PMF for all dice outcomes

dice = randint(1,7)
outcomes = list(range(1,6))

probablities = dice.pmf(outcomes)

for o, p in zip(outcomes, probablities):
    print(f"Probability of {o}: {p:.4f}")
```

```python
from scipy.stats import bernoulli

coin = bernoulli(p = 0.5)

print('Probablity of Tails:',coin.pmf(0))
print('Probablity of Head :',coin.pmf(1))
```

```python
# Suppose we roll a fair six-sided dice.
# Probability of getting 2.

die = randint(1,7)
p = die.pmf(2)
print(f"Probability of getting 2: {p}")

# Probability of getting 5.
p = die.pmf(5)
print(f"Probability of getting 2: {p}")

# Probability of getting an even number (2, 4, 6).
p = die.pmf([2,4,6]).sum()
print(f"Probability of getting an even number (2, 4, 6): {p}")

# Probability of getting an odd number (1, 3, 5).
p = die.pmf([1,3,5]).sum()
print(f"Probability of getting an odd number (1, 3, 5): {p}")
```

# CDF using SciPy

```python
# Create a dice distribution
dice = randint(1, 7)

probability = dice.cdf(4) # Calculate CDF

print("CDF:", probability)
print("Percentage:", probability * 100, "%")
```

```python
# Understand CDF with Different Values

for x in range(1,7):
    probability = dice.cdf(x)
    print(f"CDF:{x}, probablity: {probability}")
```

# PDF

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate values
x = np.linspace(-5, 5, 100)

# Calculate PDF
pdf_values = norm.pdf(x, loc=0, scale=1)

# norm represents the normal distribution.
# norm.pdf() calculates probability density values.
# loc=0 represents the mean.
# scale=1 represents the standard deviation.
# np.linspace() generates 100 equally spaced values between -5 and 5.

# Visualize
plt.plot(x, pdf_values)
plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.title("Normal Distribution PDF")
plt.grid()
plt.show()
```

# Bernoulli Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli

p = 0.6 # Here, the probability of success is 60%, and failure is 40%.

# Calculate PMF
x = [0, 1]

pmf_values = bernoulli.pmf(x, p) # Calculate probability

print(pmf_values)
```

```python
plt.bar(x, pmf_values)

plt.xticks([0, 1], ["Failure (0)", "Success (1)"])
plt.xlabel("Outcomes")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")

plt.show()
```

```python
mean = bernoulli.mean(p)
variance = bernoulli.var(p)
std = bernoulli.std(p)

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", std)
```

```python
# Generate 20 random outcomes
samples = bernoulli.rvs(p, size=20, random_state=42)
print(samples)
```

# Binomial Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Total number of trials
n = 10

# Probability of success
p = 0.5

# Calculate PMF
k = 6 # Number of successes

# Calculate probability
probability = binom.pmf(k, n, p)

print("Probability:", probability)
print("Percentage:", probability * 100)

# Calculate probabilities for all possible successes
# Possible number of successes
x = np.arange(0, n + 1)

# Calculate PMF
pmf_values = binom.pmf(x, n, p)

print(pmf_values)
```

```python
plt.bar(x, pmf_values)

plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("Binomial Distribution")

plt.xticks(x)
plt.grid(axis="y")

plt.show()
```

```python
# Mean
mean = binom.mean(n, p)

# Variance
variance = binom.var(n, p)

# Standard deviation
std = binom.std(n, p)

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", std)
```

# Normal Distribution – PDF

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu = 70    # Mean
sigma = 10 # Standard deviation

# Generate data points
x = np.linspace(30, 110, 100) # Generate values

# Calculate PDF
pdf_values = norm.pdf(x, loc=mu, scale=sigma)
print(pdf_values)

```

```python
plt.plot(x, pdf_values)

plt.xlabel("Students Marks")
plt.ylabel("Probability Density")
plt.title("Normal Distribution PDF")

plt.grid()
plt.show()
```

```python
# Calculate PDF for a specific value
pdf = norm.pdf(80, loc=70, scale=10)

print("PDF:", pdf)
```

```python
# Calculate Probability using CDF
probability = norm.cdf(80, loc=70, scale=10)

print("Probability:", probability)
print("Percentage:", probability * 100)
```

```python
# Calculate probability between two values
lower = norm.cdf(60, loc=70, scale=10)

upper = norm.cdf(80, loc=70, scale=10)

probability = upper - lower

print("Probability:", probability)
print("Percentage:", probability * 100)
```

# Z-Score in Normal Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(-4, 4, 100)

pdf = norm.pdf(x, loc=0, scale=1)

plt.plot(x, pdf)

plt.title("Standard Normal Distribution")
plt.xlabel("Z-Score")
plt.ylabel("Probability Density")

plt.grid()
plt.show()
```

```python
#  Generate Random Samples

samples = norm.rvs(
    loc=70,
    scale=10,
    size=1000,
    random_state=42
)

print(samples[:10])
```

```python
plt.hist(samples, bins=30, density=True)

plt.title("Normal Distribution - Random Samples")
plt.xlabel("Marks")
plt.ylabel("Density")

plt.show()
```

# Uniform Distribution PDF

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform

a = 10 # Minimum value
b = 20 # Maximum value

x = np.linspace(5, 25, 100) # Generate data points

# Calculate PDF
pdf_values = uniform.pdf(x, loc=a, scale=b-a)
print(pdf_values)
```

```python
plt.plot(x, pdf_values)

plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.title("Uniform Distribution PDF")

plt.grid()
plt.show()
```

```python
# Calculate PDF for a Specific Value
pdf = uniform.pdf(15, loc=10, scale=10)

print("PDF:", pdf)
```

```python
# Calculate Probability using CDF
probability = uniform.cdf(15, loc=10, scale=10)

print("Probability:", probability)
print("Percentage:", probability * 100)
```

```python
# Calculate Probability Between Two Values
lower = uniform.cdf(12, loc=10, scale=10)

upper = uniform.cdf(18, loc=10, scale=10)

probability = upper - lower

print("Probability:", probability)
print("Percentage:", probability * 100)
```

```python
# Mean
mean = uniform.mean(loc=10, scale=10)

# Variance
variance = uniform.var(loc=10, scale=10)

# Standard deviation
std = uniform.std(loc=10, scale=10)

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", std)
```

```python
samples = uniform.rvs(
    loc=10,
    scale=10,
    size=1000,
    random_state=42
)

print(samples[:10])
```

```python
plt.hist(samples, bins=20, density=True)

plt.title("Uniform Distribution - Random Samples")
plt.xlabel("Values")
plt.ylabel("Density")

plt.show()
```
