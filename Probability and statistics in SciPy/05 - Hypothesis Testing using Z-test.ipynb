# Step 1: Import libraries

```python
import numpy as np
from scipy import stats
```

# Step 2: Define the given values

```python
population_mean = 500
sample_mean = 510
population_std = 20
sample_size = 50
alpha = 0.05
```

# Step 5: Calculate the Z-statistic

```python
z_stat = (sample_mean - population_mean) / (
    population_std / np.sqrt(sample_size)
)

print("Z-statistic:", z_stat)
```

# Step 4: Calculate the P-value

```python
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print("P-value:", p_value)
```

# Step 5: Make the final decision

```python
if p_value < alpha:
    print("Reject Null Hypothesis")
    print("Average product weight is significantly different.")
else:
    print("Fail to Reject Null Hypothesis")
    print("Not enough evidence of a difference.")
```

```python
# Right-tailed
p_value = 1 - stats.norm.cdf(z_stat)

# Left-tailed
p_value = stats.norm.cdf(z_stat)

# Two-tailed
p_value = 2 * stats.norm.sf(abs(z_stat))
```
