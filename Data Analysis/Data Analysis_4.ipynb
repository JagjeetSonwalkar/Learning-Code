# Univariate on Titanic DataSet

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data_set/train.csv')
```

```python
df.head()
```

## Categorical Data

```python
df['Survived'].value_counts()
```

```python
sns.barplot(data = df['Survived'].value_counts())
```

```python
plt.pie(df['Survived'].value_counts(), autopct = '%0.1f%%')
```

## Numerical Data

```python
sns.histplot(data = df['Age'], kde=True)
```

```python
sns.distplot(df['Age'])
```

```python
sns.boxplot(df['Age'])
```
