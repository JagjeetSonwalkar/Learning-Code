```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [10, 15, 12, 18]
y2 = [5, 8, 6, 10]
y3 = [2, 4, 6, 8]
y4 = [12, 10, 15, 13]

plt.figure(figsize=(8, 6))

plt.subplot(2, 2, 1)
plt.plot(x, y1)
plt.title("Plot 1")

plt.subplot(2, 2, 2)
plt.scatter(x, y2)
plt.title("Plot 2")

plt.subplot(2, 2, 3)
plt.bar(x, y3)
plt.title("Plot 3")

plt.subplot(2, 2, 4)
plt.hist(y4)
plt.title("Plot 4")

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]

y1 = [10, 15, 12, 18]
y2 = [5, 8, 6, 10]
y3 = [2, 4, 6, 8]
y4 = [12, 10, 15, 13]

fig, ax = plt.subplots(2, 2, figsize=(10, 6))

ax[0, 0].plot(x, y1)
ax[0, 0].set_title("Line Plot")

ax[0, 1].scatter(x, y2)
ax[0, 1].set_title("Scatter Plot")

ax[1, 0].bar(x, y3)
ax[1, 0].set_title("Bar Chart")

ax[1, 1].hist(y4)
ax[1, 1].set_title("Histogram")

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

y1 = [5, 7, 6, 8, 9]
y2 = [10, 12, 11, 15, 14]

fig, ax = plt.subplots(2, 1, sharex=True)

ax[0].plot(x, y1)
ax[0].set_title("Sales")

ax[1].plot(x, y2)
ax[1].set_title("Profit")

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]

sales = [20, 25, 30, 35]
profit = [5, 7, 9, 11]

plt.plot(x, sales, label = 'sales' )
plt.plot(x, profit , label = 'profit')

plt.title("Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount")

plt.legend()

plt.show()
```
