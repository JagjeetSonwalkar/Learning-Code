```python
# Annotations
batter = pd.read_csv("DataSet/batter.csv")

sample_df = batter.head(100).sample(25, random_state = 5)
sample_df
```

```python
plt.scatter(sample_df['avg'],sample_df['strike_rate'])
plt.xlabel("avg")
plt.ylabel("strike_rate")

# from this graph we dont know which dot is a batsman, so we use annotation
```

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.scatter(x,y)

# using annotation
plt.text(1,5, 'Point 1')
plt.text(2,6, 'Point 2')
plt.text(3,7, 'Point 3')
plt.text(4,8, 'Point 4', fontdict = {'size':12,'color':'Brown'})
```

```python
plt.figure(figsize = (18,10))

plt.scatter(sample_df['avg'],sample_df['strike_rate'], s = sample_df['runs'])

plt.xlabel("avg")
plt.ylabel("strike_rate")

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i] ,sample_df['strike_rate'].values[i], sample_df['batter'].values[i])
```

```python
# plotting horizantal and vertical lines on graph.

plt.figure(figsize = (18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'], s = sample_df['runs'])
plt.xlabel("avg")
plt.ylabel("strike_rate")

plt.axhline(130, color='red')

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i] ,sample_df['strike_rate'].values[i], sample_df['batter'].values[i])
```

```python
# plotting horizantal and vertical lines on graph.

plt.figure(figsize = (18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'], s = sample_df['runs'])
plt.xlabel("avg")
plt.ylabel("strike_rate")

plt.axhline(130, color='red')
plt.axvline(30, color='blue')

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i] ,sample_df['strike_rate'].values[i], sample_df['batter'].values[i])
```

```python
# plotting horizantal and vertical lines on graph.

plt.figure(figsize = (18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'], s = sample_df['runs'])
plt.xlabel("avg")
plt.ylabel("strike_rate")

plt.axhline(130, color='red')
plt.axhline(140, color='black')
plt.axvline(30, color='blue')

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i] ,sample_df['strike_rate'].values[i], sample_df['batter'].values[i])
```

```python
# scatter plot
batter = pd.read_csv("DataSet/batter.csv")

batter
```

```python
fig = plt.figure()

ax = plt.subplot(projection='3d')
ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'])
```

```python
fig = plt.figure()
ax = plt.subplot(projection='3d')
ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'])

ax.set_title('IPL batsman analysis')

ax.set_xlabel('Run')
ax.set_ylabel('Avg')
ax.set_zlabel('SR')
```

```python
# 3D Line Plot: it is used to make the line betwwen the points
x = [0,1,5]
y = [0,10,13]
z = [0,13,20]

fig = plt.figure()

ax = plt.subplot(projection = '3d')

ax.scatter3D(x,y,z, s = [100,100,100])
ax.plot3D(x,y,z, color = 'red')
```

```python
# 3d surface plot

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx, yy = np.meshgrid(x,y)

z = xx ** 2 + yy ** 2

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection = '3d')

ax.plot_surface(xx,yy,z)
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx, yy = np.meshgrid(x,y)
z = xx ** 2 + yy ** 2
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection = '3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')

fig.colorbar(p)
```

## contour plot

```python
# 3d graph for refernce
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx, yy = np.meshgrid(x,y)
z = xx ** 2 + yy ** 2
fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection = '3d')

p = ax.plot_surface(xx,yy,z, cmap='viridis')

fig.colorbar(p)
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx, yy = np.meshgrid(x,y)
z = xx ** 2 + yy ** 2

plt.contour(x,y,z)
plt.colorbar()
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx, yy = np.meshgrid(x,y)
z = xx ** 2 + yy ** 2

plt.contourf(x,y,z)
plt.colorbar()
```
