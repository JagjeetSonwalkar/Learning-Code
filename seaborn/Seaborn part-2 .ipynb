```python
import seaborn as sns
import numpy as np
```

```python
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
```

```python
# strip plot
sns.stripplot(data = tips, x="day", y="total_bill")
```

```python
# swarmplot
sns.swarmplot(data = tips, x="day", y="total_bill")
```

```python
# box plot
sns.boxplot(data=tips, x='day', y='total_bill')
```

```python
sns.violinplot(data=tips, x='day', y = 'total_bill')
```

```python
sns.boxenplot(data=tips, x='day', y='total_bill')
```

```python
sns.barplot(tips, x = 'sex', y='total_bill', estimator=np.std)
```

```python
sns.pointplot(tips, x = 'sex', y='total_bill', estimator=np.std)
```

```python
sns.countplot(tips, x='sex')
```

```python
sns.catplot(tips, x='sex', kind='count')
```

```python
sns.regplot(data = tips, x = 'total_bill', y = 'tip')
```

```python
sns.lmplot(data = tips, x = 'total_bill', y = 'tip', hue='sex')
```

```python
sns.residplot(data = tips, x = 'total_bill', y = 'tip')
```

```python
g = sns.FacetGrid(data = tips, col = 'day', row='time')
g.map(sns.violinplot, "sex", "total_bill")
```

```python
g = sns.PairGrid(data = iris, hue="species")
g.map_diag(sns.boxplot)
g.map_offdiag(sns.scatterplot)
```

```python
sns.jointplot(data=tips, x='total_bill', y='tip')
```

```python
g = sns.JointGrid(data=tips, x='total_bill', y='tip')
g.plot(sns.scatterplot, sns.violinplot)
```
