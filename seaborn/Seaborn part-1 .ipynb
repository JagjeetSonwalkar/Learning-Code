```python
import seaborn as sns
import matplotlib as plt
import plotly.express as px    # importing for some builtin datset
```

```python
# load the dataset
tips = sns.load_dataset('tips')

tips.head()
```

```python
# scatter plot --> axes level funtion
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip')
```

```python
# relplot --> figure level funtion --> square as (default figure shape)
sns.relplot(data = tips, x = 'total_bill', y = 'tip', kind='scatter', hue = 'sex', style = 'time', size = 'size')
```

```python
flights = sns.load_dataset("flights")

sns.lineplot(data=flights, x="year", y="passengers")
```

```python
# histogram
sns.histplot(data = tips, x = 'total_bill')
```

```python
sns.displot(data = tips, x='total_bill', kind='hist')
```

```python
sns.histplot(data=tips, x="total_bill", bins=20, kde=True)
```

```python
sns.kdeplot(data=tips, x="total_bill", fill=True)
```

```python
sns.histplot(data=tips, x="total_bill", bins=20)
sns.rugplot(data=tips, x="total_bill", height=0.025, expand_margins=True, linewidth=1, c='red')
```

```python
sns.displot(data=tips, x="total_bill", kind="hist", hue="sex")
```

```python
corr = tips.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)
```

```python
corr = tips.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap="RdBu_r",
    center=0
)
```

```python
corr = tips.corr(numeric_only=True)

sns.clustermap(
    corr,
    cmap="viridis",
    annot=True
)
```
