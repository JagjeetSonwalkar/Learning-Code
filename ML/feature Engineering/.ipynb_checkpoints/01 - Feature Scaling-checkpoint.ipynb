# 5. Mean Normalization

```python
from sklearn.base import BaseEstimator, TransformerMixin
```

```python
class MeanNormalizer(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        X = np.asarray(X)
        self.mean_ = np.mean(X, axis=0)
        self.range_ = np.max(X, axis=0) - np.min(X, axis=0)
        # Avoid division by zero for constant features
        self.range_[self.range_ == 0] = 1.0
        return self

    def transform(self, X):
        X = np.asarray(X)
        return (X - self.mean_) / self.range_
```

```python
mean_scalar = MeanNormalizer()

X_train_scale = mean_scalar.fit_transform(X_train.iloc[:,4:])
X_test_scale  = mean_scalar.transform(X_test.iloc[:, 4:])
```
