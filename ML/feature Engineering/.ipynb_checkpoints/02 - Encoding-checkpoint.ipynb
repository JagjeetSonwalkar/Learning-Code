```python
import pandas as pd
import seaborn as sns
```

```python
sns.get_dataset_names()
```

## 1. One-Hot Encoding

```python
df = pd.DataFrame(sns.load_dataset('dots'))

df.sample(5)
```

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns='firing_rate')
y = pd.DataFrame(df['firing_rate'])

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

X_train
```

```python
from sklearn.preprocessing import OneHotEncoder

one_hot_encoder = OneHotEncoder(sparse_output=False)

X_train_encoder = one_hot_encoder.fit_transform(pd.DataFrame(df.iloc[:,0]))
X_test_encoder  = one_hot_encoder.transform(pd.DataFrame(df.iloc[:,0]))

X_train_encoder
```

## 2. Target Encoding

```python
df = sns.load_dataset('titanic')

df.sample(5)
```

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns='alive')
y = pd.DataFrame(df['alive'])

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

X_train
```

```python
from sklearn.preprocessing import TargetEncoder

target_encoder = TargetEncoder(smooth='auto')

X_train_encoder = target_encoder.fit_transform(pd.DataFrame(df['class']), pd.DataFrame(df['alive']))

X_train_encoder
```

## 3. Ordinal Encoding

```python
df = sns.load_dataset('flights')

df.head(5)
```

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns='passengers')
y = pd.DataFrame(df['passengers'])

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

X_train
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(
    categories = [['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']],
)

X_train_oe = oe.fit_transform(pd.DataFrame(X_train.iloc[:,1]))
X_test_oe  = oe.transform(pd.DataFrame(X_test.iloc[:,1]))

X_train_oe
```

## 4. Label Encoding

```python
df = sns.load_dataset('iris')

df.sample(5)
```

```python
from sklearn.model_selection import train_test_split

x = df.drop(columns='species')
y = pd.DataFrame(df['species'])

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

X_train
```

```python
from sklearn.preprocessing import LabelEncoder

l_encoder = LabelEncoder()

y_train_t_encoder = l_encoder.fit_transform(pd.DataFrame(y_train['species']))
y_test_t_encoder  = l_encoder.transform(pd.DataFrame(y_test['species']))

y_train_t_encoder
```
