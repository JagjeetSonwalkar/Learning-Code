# 1. Binning

```python
import numpy as np
from sklearn.preprocessing import KBinsDiscretizer

X = np.array([
    [5], [12], [18],
    [25], [35], [45], [60]
])

# Create 3 bins
kb = KBinsDiscretizer(
    n_bins=3,
    encode='ordinal',
    strategy='uniform'
)

X_binned = kb.fit_transform(X)

print(X_binned)
```

# 2. Binarization

```python
import numpy as np
from sklearn.preprocessing import Binarizer

X = np.array([
    [20], [30], [35], [60], [90]
])

binarizer = Binarizer(threshold=35)

X_binary = binarizer.transform(X)

print(X_binary)
```
