```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('Data/new_placement.csv')
```

```python
df
```

```python
plt.scatter(df.cgpa, df.package)

plt.xlabel('CGPA')
plt.ylabel('Package')
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 2)
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
# now predicting
lr.predict(X_test.iloc[0]) # Error: Beacuase the input should (1,1) not an single value
```

```python
lr.predict(X_test.iloc[0].values.reshape(1,1))
```

```python
# line
plt.scatter(df.cgpa, df.package)
plt.plot(X_train, lr.predict(X_train), color = 'red')
plt.xlabel('CGPA')
plt.ylabel('Package')
```

```python
m = lr.coef_ # slope

m
```

```python
b = lr.intercept_  # intercept

b
```

```python
# y = mx + b
y = m * 8.58 + b

y
```
