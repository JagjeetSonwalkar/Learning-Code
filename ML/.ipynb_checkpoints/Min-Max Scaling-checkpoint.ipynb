```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("Data/wine_data.csv", header = None, usecols = [0,1,2])
df.columns = ['Class label', 'Alcohol', 'Malic acid']
```

```python
df
```

```python
sns.kdeplot(df['Alcohol'])
```

```python
sns.kdeplot(df['Malic acid'])
```

```python
color_dict = {1:'red',3:'green', 2:'blue'}
sns.scatterplot(data = df, x = 'Alcohol', y = 'Malic acid', hue = df['Class label'], palette = color_dict)
```

# Train & Test Split

```python
x_train, x_test, y_train, y_test = train_test_split(df.drop('Class label', axis = 1), df['Class label'], test_size = 0.3, random_state = 0)
```

```python
print("Shape of X train:",x_train.shape)
print("Shape of Y train:",x_test.shape)
```

# Scaling

```python
scaler = MinMaxScaler()

# fit the scalar to the train set, it will learn the parameters
scaler.fit(x_train)

# transform train & test sets
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled, columns = x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled, columns = x_test.columns)
```

```python
np.round(x_train.describe(),1)
```

```python
np.round(x_train_scaled.describe(),1)
```

# Effect of Scaling

```python
fig, (ax1, ax2) = plt.subplots (ncols=2, figsize=(12,5))

ax1.scatter(x_train['Alcohol'], x_train ['Malic acid'], c = y_train)
ax1.set_title("Before Scaling")
ax2.scatter (x_train_scaled['Alcohol'], x_train_scaled['Malic acid'], c = y_train)
ax2.set_title("After Scaling")
plt.show()
```

# Comparison of Distributions

```python
fig, (ax1, ax2) = plt.subplots (ncols=2, figsize=(12,5))

# before scaling
ax1.set_title('Before Scaling')
sns.kdeplot(x_train ['Alcohol'], ax=ax1)
sns.kdeplot(x_train ['Malic acid'], ax=ax1)

# after scaling
ax2.set_title('After Standard Scaling')
sns.kdeplot(x_train_scaled ['Alcohol'], ax=ax2)
sns.kdeplot(x_train_scaled ['Malic acid'], ax=ax2)
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots (ncols=2, figsize=(12, 5))

# before scaling
ax1.set_title('Alcohol Distribution Before Scaling')
sns.kdeplot(x_train ['Alcohol'], ax=ax1)

# after scaling
ax2.set_title('Alcohol Distribution After Standard Scaling')
sns.kdeplot(x_train_scaled ['Alcohol'], ax=ax2)
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots (ncols=2, figsize=(12, 5))

# before scaling
ax1.set_title('Malic acid Distribution Before Scaling')
sns.kdeplot(x_train ['Malic acid'], ax=ax1)

# after scaling
ax2.set_title('Malic acid Distribution After Standard Scaling')
sns.kdeplot(x_train_scaled ['Malic acid'], ax=ax2)
plt.show()
```
